#### Run json file of OpenAI request trace against the OpenAI API

```python
import json
import os
from typing import Any, Dict
from openai import OpenAI

with open("prompt_playground.json", "r") as f:
    REQUEST_JSON: dict = json.load(f)

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY", "")

def pretty(obj: Any) -> str:
    return json.dumps(obj, indent=2, ensure_ascii=False)

client = OpenAI(api_key=OPENAI_API_KEY)
resp = client.chat.completions.create(**REQUEST_JSON)
# Convert the SDK object to a plain dict for printing/saving
res = json.loads(resp.model_dump_json())
```
