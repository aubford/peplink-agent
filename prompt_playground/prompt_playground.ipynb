#### Run json file of OpenAI request trace against the OpenAI API

```python
import json
import os
from typing import Any, Dict
from openai import OpenAI

with open("prompt_playground_openai.json", "r") as f:
    request: dict = json.load(f)

request["stream"] = False

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY", "")

def pretty(obj: Any):
    print(json.dumps(obj, indent=2, ensure_ascii=False))

client = OpenAI(api_key=OPENAI_API_KEY)
resp = client.chat.completions.create(**request)
# Convert the SDK object to a plain dict for printing/saving
res = json.loads(resp.model_dump_json())
```

```python
pretty(res)
```

# Make Arbitrary API calls

```python
from openai import OpenAI

client = OpenAI()

resp = client.responses.create(
    model="gpt-5",
    tools=[
        {
            "type": "mcp",
            "server_label": "dmcp",
            "server_description": "A Dungeons and Dragons MCP server to assist with dice rolling.",
            "server_url": "https://dmcp-server.deno.dev/sse",
            "require_approval": "never",
        },
    ],
    input="Roll 2d4+1",
)

print(resp)
```
